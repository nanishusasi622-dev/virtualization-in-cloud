```python
# Virtualization in Cloud Computing
# VM Resource Allocation Simulation
# Google Colab

import random
import matplotlib.pyplot as plt

# Physical server resources
TOTAL_CPU = 100       # CPU capacity in %
TOTAL_MEMORY = 16000  # Memory capacity in MB

# Number of Virtual Machines
num_vms = 5

# Generate virtual machines
vms = []

for i in range(1, num_vms + 1):
    cpu = random.randint(10, 30)
    memory = random.randint(1000, 4000)

    vms.append({
        "VM": f"VM-{i}",
        "CPU": cpu,
        "Memory": memory
    })

# Display physical server information
print("VIRTUALIZATION IN CLOUD COMPUTING")
print("=" * 60)

print(f"Physical CPU Capacity    : {TOTAL_CPU}%")
print(f"Physical Memory Capacity : {TOTAL_MEMORY} MB")
print(f"Number of Virtual Machines: {num_vms}")

print("\nVIRTUAL MACHINE RESOURCE ALLOCATION")
print("=" * 60)

total_cpu = 0
total_memory = 0

# Allocate resources
for vm in vms:
    total_cpu += vm["CPU"]
    total_memory += vm["Memory"]

    print(
        f"{vm['VM']} -> "
        f"CPU: {vm['CPU']}% | "
        f"Memory: {vm['Memory']} MB"
    )

print("\n" + "=" * 60)

# Check whether physical resources are sufficient
if total_cpu <= TOTAL_CPU and total_memory <= TOTAL_MEMORY:
    print("Status: All Virtual Machines allocated successfully.")
else:
    print("Status: Resource limit exceeded.")

# Resource utilization
cpu_utilization = (total_cpu / TOTAL_CPU) * 100
memory_utilization = (total_memory / TOTAL_MEMORY) * 100

print(f"\nTotal CPU Allocated    : {total_cpu}%")
print(f"Total Memory Allocated : {total_memory} MB")
print(f"CPU Utilization        : {cpu_utilization:.2f}%")
print(f"Memory Utilization     : {memory_utilization:.2f}%")

# VM names
vm_names = [vm["VM"] for vm in vms]

# CPU usage graph
plt.figure(figsize=(8, 5))
plt.bar(vm_names, [vm["CPU"] for vm in vms])
plt.xlabel("Virtual Machines")
plt.ylabel("CPU Allocation (%)")
plt.title("CPU Allocation to Virtual Machines")
plt.show()

# Memory usage graph
plt.figure(figsize=(8, 5))
plt.bar(vm_names, [vm["Memory"] for vm in vms])
plt.xlabel("Virtual Machines")
plt.ylabel("Memory Allocation (MB)")
plt.title("Memory Allocation to Virtual Machines")
plt.show()
```
